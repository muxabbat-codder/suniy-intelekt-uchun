```python
import numpy as np

# Leaky Integrate-and-Fire (LIF) Neuron Model
class LIFNeuron:
    def __init__(self, threshold, reset_value, decay_factor, refractory_period):
        self.threshold = threshold
        self.reset_value = reset_value
        self.decay_factor = decay_factor
        self.refractory_period = refractory_period
        self.membrane_potential = 0
        self.refractory_end_time = -1
        self.last_spike_time = -np.inf # Added for STDP: Initialize to a very old time

    def update(self, incoming_current, current_time): # Renamed `incoming_spikes` to `incoming_current` for clarity
        if current_time < self.refractory_end_time:
            # Neuron is in refractory period, cannot spike
            return False

        self.membrane_potential = (self.membrane_potential * self.decay_factor) + incoming_current

        if self.membrane_potential >= self.threshold:
            self.membrane_potential = self.reset_value
            self.refractory_end_time = current_time + self.refractory_period
            self.last_spike_time = current_time # Record spike time
            return True
        return False

# Synapse and STDP Learning Rule
class Synapse:
    def __init__(self, weight): self.weight = weight

def stdp(pre_time, post_time, weight, lr, tau_p, tau_n):
    # Only apply STDP if both pre- and post-synaptic neurons have spiked at a finite time
    if pre_time != -np.inf and post_time != -np.inf:
        dt = post_time - pre_time
        if dt > 0: # LTP (Long-Term Potentiation)
            return weight + (lr * np.exp(-dt / tau_p))
        elif dt < 0: # LTD (Long-Term Depression)
            return weight - (lr * np.exp(dt / tau_n)) # Use '-' for depression
    return weight # No change if only one or neither spiked, or dt == 0

# Simulation Parameters
time_steps = 100
input_size = 5
hidden_size = 3
output_size = 1

# Network Initialization
input_neurons = [LIFNeuron(threshold=1.0, reset_value=0.0,
decay_factor=0.9, refractory_period=2) for _ in range(input_size)]
hidden_neurons = [LIFNeuron(threshold=1.0, reset_value=0.0,
decay_factor=0.9, refractory_period=2) for _ in range(hidden_size)]
output_neurons = [LIFNeuron(threshold=1.0, reset_value=0.0,
decay_factor=0.9, refractory_period=2) for _ in range(output_size)]

input_to_hidden_synapses = np.random.rand(input_size, hidden_size)
hidden_to_output_synapses = np.random.rand(hidden_size, output_size)

learning_rate = 0.01
tau_positive = 20
tau_negative = 20

# Spike Train Pattern to Detect
pattern = [1, 0, 1, 0, 1]

# Simulation Loop
for t in range(time_steps):
    # Generate external input for the input layer (0 or 1 for spike presence)
    current_external_input = np.random.randint(0, 2, size=input_size)

    # Track which neurons fired in each layer for spike propagation and STDP
    fired_input_neurons_indices = []
    fired_hidden_neurons_indices = []

    # 1. Update Input Neurons
    # Each input neuron receives its corresponding external_input.
    for i, neuron in enumerate(input_neurons):
        if neuron.update(current_external_input[i], t):
            fired_input_neurons_indices.append(i)

    # 2. Calculate incoming current for Hidden Neurons from fired Input Neurons
    incoming_current_to_hidden = np.zeros(hidden_size)
    for i in fired_input_neurons_indices:
        # If input neuron 'i' fired, it sends weighted spikes to all hidden neurons.
        incoming_current_to_hidden += input_to_hidden_synapses[i, :]

    # 3. Update Hidden Neurons
    for j, neuron in enumerate(hidden_neurons):
        if neuron.update(incoming_current_to_hidden[j], t):
            fired_hidden_neurons_indices.append(j)

    # 4. Calculate incoming current for Output Neurons from fired Hidden Neurons
    incoming_current_to_output = np.zeros(output_size)
    for j in fired_hidden_neurons_indices:
        # If hidden neuron 'j' fired, it sends weighted spikes to all output neurons.
        incoming_current_to_output += hidden_to_output_synapses[j, :]

    # 5. Update Output Neurons
    for k, neuron in enumerate(output_neurons):
        neuron.update(incoming_current_to_output[k], t)

    # STDP Learning
    # Apply STDP to input-to-hidden synapses
    for i in range(input_size):
        for j in range(hidden_size):
            input_to_hidden_synapses[i, j] = stdp(input_neurons[i].last_spike_time,
                                                 hidden_neurons[j].last_spike_time,
                                                 input_to_hidden_synapses[i, j],
                                                 learning_rate, tau_positive, tau_negative)

    # Apply STDP to hidden-to-output synapses
    for j in range(hidden_size):
        for k in range(output_size):
            hidden_to_output_synapses[j, k] = stdp(hidden_neurons[j].last_spike_time,
                                                  output_neurons[k].last_spike_time,
                                                  hidden_to_output_synapses[j, k],
                                                  learning_rate, tau_positive, tau_negative)

    # Check if pattern is detected
    # The pattern is defined over the input layer.
    # Check if all input neurons corresponding to '1' in the pattern fired at current time 't'.
    pattern_matched_at_t = True
    for p_idx, p_val in enumerate(pattern):
        if p_val == 1: # If the pattern expects a spike from this input neuron
            if p_idx not in fired_input_neurons_indices: # Check if it actually fired
                pattern_matched_at_t = False
                break
        # If p_val is 0, we don't explicitly enforce it not to fire for pattern detection
        # as per the original intent (only checking '1's).

    # Ensure there's at least one '1' in the pattern for a meaningful detection
    if pattern_matched_at_t and any(p == 1 for p in pattern):
        print(f"Pattern detected at time step {t}")
```
