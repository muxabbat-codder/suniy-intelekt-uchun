```python
import numpy as np

# --- Классы нейронов и STDP (Изображение 1) ---
class LIFNeuron:
    def __init__(self, threshold, reset_value, decay_factor, refractory_period):
        self.threshold = threshold
        self.reset_value = reset_value
        self.decay_factor = decay_factor
        self.refractory_period = refractory_period
        self.membrane_potential = 0
        self.spike_time = -1
        self.refractory_end_time = -1

    def update(self, incoming_spikes, current_time):
        if current_time < self.refractory_end_time: return False
        self.membrane_potential = self.membrane_potential * self.decay_factor + np.sum(incoming_spikes)
        if self.membrane_potential >= self.threshold:
            self.spike_time, self.membrane_potential = current_time, self.reset_value
            self.refractory_end_time = current_time + self.refractory_period
            return True
        return False

def stdp(pre_spike, post_spike, weight, lr, tau_pos, tau_neg):
    if pre_spike > 0 and post_spike > 0:
        dt = post_spike - pre_spike
        return weight + (lr * np.exp(-dt / tau_pos) if dt > 0 else -lr * np.exp(dt / tau_neg))
    return weight

# --- Инициализация (Изображение 2) ---
time_steps, input_size, hidden_size, output_size = 100, 5, 3, 1
lr, tau_pos, tau_neg = 0.01, 20, 20
input_neurons = [LIFNeuron(1.0, 0.0, 0.9, 2) for _ in range(input_size)]
hidden_neurons = [LIFNeuron(1.0, 0.0, 0.9, 2) for _ in range(hidden_size)]
output_neurons = [LIFNeuron(1.0, 0.0, 0.9, 2) for _ in range(output_size)]
syn_in_hid = np.random.rand(input_size, hidden_size)
syn_hid_out = np.random.rand(hidden_size, output_size)

# --- Цикл симуляции (Изображение 3) ---
for t in range(time_steps):
    in_spikes = np.random.randint(0, 2, size=input_size)
    hid_spikes = np.zeros(hidden_size)
    # Обновление нейронов (input -> hidden -> output)
    for i, n in enumerate(input_neurons):
        if n.update(in_spikes[i] * syn_in_hid[i], t): hid_spikes += syn_in_hid[i]
    out_spikes = np.zeros(output_size)
    for j, n in enumerate(hidden_neurons):
        if n.update(hid_spikes[j] * syn_hid_out[j], t): out_spikes += syn_hid_out[j]
    for k, n in enumerate(output_neurons): n.update(out_spikes[k], t)

    # STDP Обучение
    for i in range(input_size):
        for j in range(hidden_size):
            syn_in_hid[i, j] = stdp(input_neurons[i].spike_time, hidden_neurons[j].spike_time, syn_in_hid[i, j], lr, tau_pos, tau_neg)
    for j in range(hidden_size):
        for k in range(output_size):
            syn_hid_out[j, k] = stdp(hidden_neurons[j].spike_time, output_neurons[k].spike_time, syn_hid_out[j, k], lr, tau_pos, tau_neg)
```
